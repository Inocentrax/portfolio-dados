# Divisão dados treino e teste

# Utilizando biblioteca para dividir treino e teste

```python
from sklearn.model_selection import train_test_split
train_test_split(x,y)
```

* Dividiu em 4 pedaços
1. treino x
2. treino y
3. teste x
4. teste y

Documentação do Train & Teste Split, tem Ramdom State
Se não for aleatorio será pré definido

* Note que se rodar o trecho a baixo a acuracia sempre se alterará

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

treino_x, teste_x, treino_y, teste_y = train_test_split(x,y)

print(f'Treinaremos com {len(treino_x)} elementos')
print(f'testaremos com {len(teste_x)} elementos')

modelo = LinearSVC()
modelo.fit(treino_x,treino_y)
previsoes = modelo.predict(teste_x)

acuracia = accuracy_score(teste_y,previsoes) * 100
print(f'A acuracia foi de {acuracia:.2f}%')
```

Reduzindo aleatoriedade

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

SEED= 246 #colocar numero aleatorio
treino_x, teste_x, treino_y, teste_y = train_test_split(x,y, random_state=SEED)

print(f'Treinaremos com {len(treino_x)} elementos')
print(f'testaremos com {len(teste_x)} elementos')

modelo = LinearSVC()
modelo.fit(treino_x,treino_y)
previsoes = modelo.predict(teste_x)

acuracia = accuracy_score(teste_y,previsoes) * 100
print(f'A acuracia foi de {acuracia:.2f}%')
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

SEED= 246 #colocar numero aleatorio
treino_x, teste_x, treino_y, teste_y = train_test_split(x,y, random_state=SEED, stratify=y)

print(f'Treinaremos com {len(treino_x)} elementos')
print(f'testaremos com {len(teste_x)} elementos')

modelo = LinearSVC()
modelo.fit(treino_x,treino_y)
previsoes = modelo.predict(teste_x)

acuracia = accuracy_score(teste_y,previsoes) * 100
print(f'A acuracia foi de {acuracia:.2f}%')
```
