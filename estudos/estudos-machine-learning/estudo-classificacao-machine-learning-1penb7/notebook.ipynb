# Implementação de Estimador

![image.png](assets/figura-07.png)

## testando a logica -TREINO DE MODELO

```python
#Features: 1 = SIM / 0 = NÃO
#Pelo longo / Perna Curta / Faz auau

porco1 = [0,1,0]#não, sim, não
porco2 = [0,1,1]#não, sim, sim
porco3 = [1,1,0]#sim, sim, não

cachorro1 = [0,1,1]#não, sim, não
cachorro2 = [1,0,1]#sim, não, sim
cachorro3 = [1,1,1]#sim, sim, sim

treino_X = [porco1, porco2, porco3, cachorro1, cachorro2, cachorro3]#featuress
treino_Y = [1,1,1,0,0,0] #1=porco / 0=cachorro
```

* Criaremos um modelo que prevê essas classes.
* Imagine grandes volumes de dados(Nossa base de dados é introdutória(Por enquanto))


```python
from sklearn.svm import LinearSVC
model = LinearSVC() #tipo de modelo
model.fit(treino_X, treino_Y) #treinar com dados baseados nas respectivas classes(Porcos e cães)
```

## teste com modelo ja treinado

```python
#pelo longo(1=sim),perna curta(1=sim), faz auau (1=ssim)
teste = [1,1,1]#cachorro
model.predict([teste])
```

```python
#pelo longo(0=não),perna curta(0=não), faz auau (0=não)
teste = [0,0,0]#porco
model.predict([teste])
```

* lembrando 0 = cão
* lembrando 1 = porco

* como temos poucos exemplos e poucas caracteristicas fica facil prever.

## criação de dados novos PARA TESTAR

```python
#Testando quantos ele acerta.
misterio1 = [1,1,1]#sim, sim, sim
misterio2=[1,1,0]#sim, sim, não
misterio3=[0,1,1]#não, sim, sim
```

```python
teste_X = [misterio1, misterio2, misterio3]
```

```python
previsoes = model.predict(teste_X)#usando modelo para prever os testes
```

```python
print(previsoes)#SAIDA = cao, porco, cao - acertou 2
```

-  mas o importante é a porcentagem de chance que ele tem de acertar, comumente chamado de (NOTA)

```python
teste_Y = [0,1,1]#cão, porco, porco - CLASSES CORRETAS PARA COMPARAR COM ACERTOS E ERROS DO TESTE
```

```python
corretos = (previsoes == teste_Y).sum()#comparar previsões com as classes que ditamos e ver a porcentagem de acerto
```

```python
total = len(teste_X)#em testes utilizamos 3 itens, o len será 3
```

```python
taxa_de_acerto = corretos / total *100 # se corretos é 2, e total é 3 * 6 == 66%
```

```python
print(f'Acurácia:{taxa_de_acerto:.2f}%')
```

```python
from sklearn.metrics import accuracy_score
```

```python
taxa_de_acerto = accuracy_score(teste_Y, previsoes)*100
print(f'{taxa_de_acerto:.2f}%')
```
