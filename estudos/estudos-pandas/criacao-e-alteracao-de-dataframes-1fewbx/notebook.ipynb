## Desenvolvimento


```python
import pandas as pd
```

```python
dados = {"Nome":["Ana", "Pedro", "Paulo", "Matheus"],
         "Idade":[25,28,24,26],
         "Cidade":["São Paulo", "Curitiba", "Rio de Janeiro", "São Sebastião"]}
```

```python
# O objeto inicial é um dicionário; info() será usado após pd.DataFrame.
type(dados)
```

```python
dados = pd.DataFrame(dados)
dados
```

```python
dados.head(2)
```

```python
dados.tail(2)
```

```python
dados["Cidade"]
```

```python
dados.isnull().sum()
```

```python
dados.loc[len(dados)] = ["Tais", 26, "São sebastião"]
```

```python
dados
```

```python
dados
```

```python
dados.insert(1, "Sexo", ["Feminino","Masculino", "Masculino","Masculino","Feminino"])
```

```python
dados
```

```python
dados.insert(4, "Sigla", ["SP","CU","RJ","BSB","BSB"])
```

```python
dados
```

```python
dados.drop("Sigla", axis=1, inplace=True)
```

```python
dados
```

```python
dados.boxplot("Idade", by="Sexo")
```
