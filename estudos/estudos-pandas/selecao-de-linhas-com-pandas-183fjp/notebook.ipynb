## Desenvolvimento


```python
import pandas as pd
```

```python
dados = ({
    'Nome':["Peixoto","Cabra-Peste","Junin","Rodolfo","Pitanga"],
    'Sobrenome':["Barroso","Chupisco", "Inverno", "Patati", "Patata"],
    'Idade':[12,13,15,11,17],
    'Altura':[2.20,1.85,1.62,1.55,2.10]
})
dados
```

```python
dados = pd.DataFrame(dados)
dados.head()
```

```python
dados.nlargest(5,'Altura')
```

```python
dados.nlargest(5, "Idade")
```
