```python
def validar(agente, recurso):
  if agente in permissoes:
    if permissoes[agente]==recurso:
      return 'Recurso autorizado, acesso autorizado'
    else:
      return 'Recurso negado, acesso negado'
  else:
    return 'agente desconhecido, tente novamente'

resultado = validar('auditor', 'logs_auditoria')
print(resultado)
```

```python
agente= 'auditor'
recurso='logs_auditoria'

print(validar(agente, recurso))
```
