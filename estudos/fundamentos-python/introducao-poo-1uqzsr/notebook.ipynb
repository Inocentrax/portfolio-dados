# Bicicletaria.
* João tem uma bicicletaria e gostaria de registrar as vendas de suas bicicletas.

* Crieu um programa onde João informe: cor, modelo, ano e valor da bicicleta vendida.

* Uma bicicleta pode:
buxinar, parar e correr. adicione esses comportamentos!

```python
class Bicicleta:
  def __init__ (self, cor, modelo, ano, valor):
    self.cor = cor
    self.modelo = modelo
    self.ano = ano
    self.valor = valor

  def Buzinar(self):
    print('Piiiiii')

  def Parar(self):
    print('Parando Bicicleta')
    print('Bicicleta parada!')

  def Correr(self):
    print('Vrummmmmm')
```

* self é a instancia do objeto.
* Outras linguagens são implicitas, no python é explicito.

* Parar, Correr e Buzinar, São conduzidos por metodos.



```python
b1 = Bicicleta('Vermelha', 'Caloi 10', 2022, 600)
```

```python
b1.Buzinar()
```

```python
b1.Parar()
```

# Poliformismo

Mesmo metodo, comportamento diferente

```python
nome = 'python'
lista = [1,2,3]
print(len(nome))
print(len(lista))
```

polimorfismo com herança herda atributos da classe pai.
É possivel modificar metodos em uma classe filha herdade da classe pai(Util quando o metodo herdado da classe pai não se encaixa perfeitamente na classe filha)

```python
class Passaro:
  def voar(self):  #classe pai
    print('voando')

class Pardal(Passaro):#classe filho
  def voar(self):
    print('pardal voando')

class Avestruz(Passaro):
  def voar(self):
    print('Avestruz não voa')

def plano_de_voo(obj):
  obj.voar()


class Aviao(Passaro):#extensão do comportamento de voar
  def voar(self):
    print('Avião está voando!')
plano_de_voo(Pardal())
plano_de_voo(Avestruz())
plano_de_voo(Aviao())
```

Tenho um Objeto e ele se comporta de formas diferentes chamando o nome(metodo).
