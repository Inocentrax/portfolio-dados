## Desenvolvimento


```python
lista = [10, 20, 30, 40, 50]

n = len(lista)

for j in range(n // 2):
    temp = lista[j]
    lista[j] = lista[n - j - 1]
    lista[n - j - 1] = temp

print(lista)
```

```python
lista_nomes = ['Amanda','Bernardo','Carlos','Daniel','Ednaldo']
tamanho_lista = len(lista_nomes)

for posicao in range(tamanho_lista//2): #Dividir lista em 2 partes
  temporaria = lista_nomes[posicao] #Temporaria=index
  lista_nomes[posicao] = lista_nomes[tamanho_lista-posicao-1]#ultima para frente
  lista_nomes[tamanho_lista-posicao-1] = temporaria#primeira para ultima

print(lista_nomes)
```
