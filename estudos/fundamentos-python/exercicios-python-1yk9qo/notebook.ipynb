```python
trocas
```

```python
trocas = [5, 10, 15, 20, 25]

for i in range(len(trocas)//2):
  outro_indice = len(trocas) -1 - i
  temporario = trocas[i]
  trocas[i] = trocas[outro_indice]
  trocas[outro_indice]=temporario

trocas
```
