## Desenvolvimento


A função deve realizar as seguintes operações:

1- Verificar se a imagem possui mais de um canal. Caso possua três canais (RGB), convertê-la para um único canal, calculando a média dos três canais para cada pixel. (Vereficação de canais e calculo de cada canal)

2- Aplicar o filtro recebido como parâmetro sobre a imagem resultante. Considere que o filtro é uma matriz NumPy de mesmas dimensões da imagem (H,W), e que a operação consiste na multiplicação elemento a elemento entre a imagem e o filtro. Cheque se o filtro tem a mesma dimensão caso contrário retorne a imagem original.(Se tem a mesma dimensão, aplica, senão imagem original, conferencia de dimensão)

3- Garanta que, após a aplicação do filtro, todos os valores dos pixels permaneçam no intervalo [0, 255].(limitação de pixels)

4- Retornar a imagem processada (ou imagem original se o filtro não tiver mesma dimensão H,W da imagem).

# Modificações aplicadas

```python
#Estrutura da função


import numpy as np


def aplicar_filtro(imagem, filtro):

    #se imagem possuir três canais (RGB), calcula media de três canais para cadapixel.
    if imagem.ndim == 3:
        imagem_processada = np.mean(imagem, axis=2)
    else:
        imagem_processada = imagem.copy()

    #se o filtro possui as mesmas dimensões de H e W.
    #se não, retorna a imagem original.
    if filtro.shape != imagem_processada.shape:
        return imagem


    imagem_processada = imagem_processada * filtro #Multiplica cada valor da imagem pelo valor do filtro

    imagem_processada = np.clip(imagem_processada, 0, 255) #valores entre 0 e 255

    imagem_processada = imagem_processada.astype(int) #valores para inteiros

    return imagem_processada


#Exemplos de execução

# Exemplo 1
imagem = np.array([[100,200],[50,250]])
filtro = np.array([[1.0,0.5],[2.0,1.2]])
resultado = aplicar_filtro(imagem, filtro)
print(f"{resultado}  \n")


# Exemplo 2
imagem = np.array([
    [[100, 150, 200], [200, 100, 60]],
    [[50, 100, 150], [240, 210, 180]]
])
filtro = np.array([[1.0, 0.5], [2.0, 1.2]])
resultado = aplicar_filtro(imagem, filtro)
print(f"{resultado}  \n")


# Exemplo 3
imagem = np.array([[100, 200],[50, 250]])
filtro = np.array([[1.0],[2.0]])
resultado = aplicar_filtro(imagem, filtro)
print(f'{resultado} ')
```
