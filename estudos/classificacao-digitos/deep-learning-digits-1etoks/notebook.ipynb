## Desenvolvimento


Carregamento de dados

```python
from sklearn.datasets import load_digits

digits = load_digits()
```

visualização de dados

```python
print(digits.data.shape)
```

tipo de dados

```python
print(type(digits.data))
print(digits.data.dtype)
```

divisão de dados

```python
X = digits.data
y = digits.target
```

* Visualização X - Item
* Visualização Y - Pixels

```python
print("Primeira resposta correta:", y[0])
print("Quantidade de pixels da primeira imagem:", len(X[0]))
```

Organizando primeira imagem

```python
primeira_imagem = X[0].reshape(8, 8)

print(primeira_imagem)
print(primeira_imagem.shape)
```

Visualização primeira imagem

```python
import matplotlib.pyplot as plt
plt.imshow(primeira_imagem, cmap="gray")
```

Tom do pixel

```python
print(primeira_imagem[0, 0])#escuro
```

```python
print(primeira_imagem[0, 3])#claro
```

Normalizando valor de pixel

```python
pixel_normalizado = primeira_imagem[0, 3] / 16

print(pixel_normalizado)
```

```python
X_normalizado = X / 16.0
```

```python
print(X[0, 3])
print(X_normalizado[0, 3])
print(X_normalizado.shape)
```

Divisão de dados treino e teste tanto de X quanto de Y

```python
from sklearn.model_selection import train_test_split
```

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(X_normalizado,y,test_size=0.2,random_state=42,stratify=y)
```

```python
print("X treino:", X_treino.shape)
print("X teste:", X_teste.shape)
print("y treino:", y_treino.shape)
print("y teste:", y_teste.shape)
```

Organização tensor X

```python
import torch
```

```python
X_treino_tensor = torch.tensor(X_treino, dtype=torch.float32)
```

```python
print(X_treino_tensor.shape)
print(X_treino_tensor.dtype)
```

Organizando Tensor Y

```python
y_treino_tensor = torch.tensor(y_treino, dtype=torch.long)
```

```python
print(y_treino_tensor.shape)
print(y_treino_tensor.dtype)
print(y_treino_tensor[:5])
```

Camada rede neural

```python
camada = torch.nn.Linear(64, 10)
```

```python
saida = camada(X_treino_tensor[:1])
print(saida.shape)
```

```python
print(saida)
```

previsão primeiro item comparando com valor real

```python
previsao = torch.argmax(saida, dim=1)
print("Número previsto:", previsao.item())
```

```python
previsao = torch.argmax(saida, dim=1)

print("Número previsto:", previsao.item())
print("Número correto:", y_treino_tensor[0].item())
```

```python
funcao_perda = torch.nn.CrossEntropyLoss()
```

Calculo de função de erro sem conhecimento de itens

```python
erro = funcao_perda(saida, y_treino_tensor[:1])

print("Perda:", erro.item())
```

Usa o erro

```python
erro.backward()
```

```python
print(camada.weight.grad.shape)
```

melhora o erro

```python
otimizador = torch.optim.SGD(camada.parameters(), lr=0.01)
```

```python
peso_antes = camada.weight[0, 0].item()

print("Peso antes:", peso_antes)
```

```python
otimizador.step()
```

```python
peso_depois = camada.weight[0, 0].item()

print("Peso antes:", peso_antes)
print("Peso depois:", peso_depois)
```

Otimização de paremetros com redução de erro

```python
print("Pixel:", X_treino_tensor[0, 3].item())
print("Gradiente:", camada.weight.grad[0, 3].item())
```

```python
nova_saida = camada(X_treino_tensor[:1])

novo_erro = funcao_perda(nova_saida, y_treino_tensor[:1])

print("Perda antes:", erro.item())
print("Perda depois:", novo_erro.item())
```

```python
nova_previsao = torch.argmax(nova_saida, dim=1)

print("Número previsto:", nova_previsao.item())
print("Número correto:", y_treino_tensor[0].item())
```

```python
saidas_5 = camada(X_treino_tensor[:5])

previsoes_5 = torch.argmax(saidas_5, dim=1)

print("Previsões:", previsoes_5)
print("Respostas: ", y_treino_tensor[:5])
```

```python
acertos = (previsoes_5 == y_treino_tensor[:5])

print(acertos)
```

```python
acuracia = acertos.float().mean()
print(acuracia)
```

```python
print("Acurácia:", acuracia.item() * 100, "%")
```
