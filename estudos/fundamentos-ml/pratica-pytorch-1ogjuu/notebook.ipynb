## Desenvolvimento


```python
dados = {
    'Horas':[1,2,3,4,5,6],
    'Consumo':[5,8,11,14,17,20]
}
```

```python
import torch
```

```python
x = torch.tensor(dados['Horas'], dtype=torch.float32)
y = torch.tensor(dados['Consumo'], dtype=torch.float32)
```

```python
print(x)
print(y)
print(x.shape)
print(y.shape)
```

```python
print(f"Antes: {x.shape}")
```

```python
x = x.reshape(6,1)
print(f'depois: {x.shape}')
```

```python
from torch import nn
```

```python
modelo = nn.Linear(1,1)
```

observar seus parâmetros iniciais

```python
print("Peso:", modelo.weight)
print("Bias:", modelo.bias)
```

PyTorch cria automaticamente um peso (w) e um bias (b) com valores iniciais.

```python
previsoes = modelo(x)
print(previsoes)
```

```python
funcao_perda = nn.MSELoss()
```

```python
y_real = y.reshape(-1,1)
print(y_real)
```

```python
print(f'X :{x.shape}')
print(f'Y: {y_real.shape}')
```

```python
erro = funcao_perda(previsoes, y_real)
print(erro.item())
```
