# Coeficiente de Correlação de Pearson.

```python
import pandas as pd

df = {
"Pessoas":[1,2,3,4,5],
"Horas":[1,2,3,4,5],
"Notas":[52, 60, 65, 80, 85]
}

X = df['Horas']
Y = df['Notas']
```

```python

i = 0
soma_notas = 0
qtd_horas = 0
qtd_pessoas = 0

while i < len(df['Notas']):
  soma_notas+=df['Notas'][i]
  qtd_horas+=df['Horas'][i]
  qtd_pessoas+=1
  i+=1

print(soma_notas)
print(qtd_horas)
print(qtd_pessoas)
```

1. qual é o valor médio das horas estudadas (X)
2. qual é o valor médio notas obtidas (Y).

```python
Media_X = qtd_horas / qtd_pessoas   # Média de X
Media_Y = soma_notas / qtd_pessoas  # Média de Y

print(Media_X) # x barra
print(Media_Y)# y barra
```

calculamos os desvios em relação à média
1. Para cada aluno, calculamos a diferença entre seu valor e a média.
2. são desvios individuais.

```python
desvio_Y = []
desvio_X = []
i = 0

while i < len(df['Horas']):
  desvio_X.append(df['Horas'][i] - Media_X ) # Desvio de X
  desvio_Y.append(df['Notas'][i] - Media_Y)  # Desvio de Y
  i+=1

df['Desvio_X'] = desvio_X
df['Desvio_Y'] = desvio_Y

print(desvio_Y)
print(desvio_X)
```


1.   Multiplicamos os desvios de X e Y ( soma_produto )
2.   Elevamos o desvio de X ao quadrado ( soma_quadratica_X )
3.   Elevamos o desvio de Y ao quadrado ( soma_quadratica_Y )
4.   Obtemos três resultados finais(soma_produto, total_soma_quadratica_X, total_soma_quadratica_Y )




```python
i = 0
produto = []
soma_produto = 0
soma_quadratica_X = []
soma_quadratica_Y = []
total_soma_quadratica_Y = 0
total_soma_quadratica_X = 0

while i < len(df['Horas']):
  produto.append(desvio_Y[i] * desvio_X[i])           # Produto dos desvios de X e Y.
  soma_produto+=produto[i]                            # Soma de produtos ao quadrado.
  soma_quadratica_X.append(desvio_X[i]*desvio_X[i])   # Guarda o quadrado do desvio de X
  soma_quadratica_Y.append(desvio_Y[i] * desvio_Y[i]) # Guarda o quadrado do desvio de Y
  total_soma_quadratica_X+=soma_quadratica_X[i]       # Soma os quadrados de X
  total_soma_quadratica_Y+=soma_quadratica_Y[i]       # Soma os quadrados de Y
  i+=1

print(produto)
print(soma_produto)
print(soma_quadratica_X)
print(soma_quadratica_Y)
print(f'Total da soma quadratica de Y: {total_soma_quadratica_Y},Total da soma quadratica de X:{total_soma_quadratica_X}')
```

1. utilizando a soma dos quadrados das notas.

```python
import math as mt

raiz_soma_quadratica_X = mt.sqrt(total_soma_quadratica_X) # Raiz quadrada de X
raiz_soma_quadratica_Y = mt.sqrt(total_soma_quadratica_Y) # Raiz quadrada de Y

print(f'Raiz quadrada de soma quadraticaY: {raiz_soma_quadratica_Y}')
print(f'Raiz quadrada de soma quadraticaX: {raiz_soma_quadratica_X}')
```

 Calcular o denominador de Pearson.
1. raiz_soma_quadratica_X
2. raiz_soma_quadratica_Y

```python
denominador = raiz_soma_quadratica_Y * raiz_soma_quadratica_X
print(denominador)
```

 resultado final da correlação de Pearson.
1. Produto
2. Denominador

```python
r_Person = soma_produto / denominador
print(r_Person)
```

* + → quando X aumenta, Y tende a aumentar.
*  − → quando X aumenta, Y tende a diminuir.

1. Próximo de 0 → relação linear fraca.
2. Próximo de −1 ou +1 → relação linear forte.

* O que acontece se aluno estudarem 6 horas?

1. ~y = ax + b = Na regressão linear simples, utilizamos a equação de uma reta:

* ~y	- Nota prevista pelo modelo
* a	- Inclinação da reta: quanto a previsão muda a cada hora adicional
* b	- Intercepto: previsão quando X é zero
* x	- Horas estudadas


* Diferença nas horas
* Diferença nas notas
* dividimos Diferença nas notas/Diferença nas horas

```python
a = soma_produto / total_soma_quadratica_X

print(a)
```

Calculando o intercepto (b)
* b= ~y-a ~x

```python
b = Media_Y - (a * Media_X)
print(b)
```

```python
x_exemplo = 1
y = a * x_exemplo + b
print(y)
```

```python

import numpy as np
import matplotlib.pyplot as plt

#Produzir uma sequência reproduzível
np.random.seed(42)

# Horas de estudo
# 30 valores entre 0 e 10
X = np.linspace(0, 10, 30)
# Notas simuladas com um padrão não linear
y = 20 + 12 * X - 0.9 * X**2
y = y + np.random.normal(0, 7, 30)

plt.scatter(X, y)
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.title("Dados de treinamento")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_treino, X_validacao, y_treino, y_validacao = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Dados de treinamento:", len(X_treino))
print("Dados de validação:", len(X_validacao))
```

```python

from sklearn.linear_model import LinearRegression

modelo_linear = LinearRegression()

modelo_linear.fit(X_treino.reshape(-1, 1), y_treino)
```

```python

print("Peso (w):", modelo_linear.coef_[0])
print("Bias (b):", modelo_linear.intercept_)
```

Exemplo 5 Horas

```python
w = modelo_linear.coef_[0]
b = modelo_linear.intercept_

y_barra = w*(5)+b

#previsão de nota
print(y_barra)
```

```python
plt.scatter(X, y, label="Dados simulados")

y_previsto = modelo_linear.predict(X.reshape(-1, 1))

plt.plot(X, y_previsto, color="red", label="Modelo linear")

plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.title("Dados simulados x Modelo linear")
plt.legend()
plt.show()
```

```python
print("Quantidade de X:", len(X))
print("Quantidade de y:", len(y))
print("Formato de X:", X.shape)
print("Formato de y:", y.shape)
```

```python

from sklearn.metrics import mean_squared_error

# Previsões nos dados conhecidos
previsao_treino = modelo_linear.predict(
    X_treino.reshape(-1, 1)
)

# Previsões nos dados reservados
previsao_validacao = modelo_linear.predict(
    X_validacao.reshape(-1, 1)
)

# Erro médio quadrático
mse_treino = mean_squared_error(y_treino, previsao_treino)
mse_validacao = mean_squared_error(y_validacao, previsao_validacao)

print("MSE treino:", mse_treino)
print("MSE validação:", mse_validacao)
```

```python

from sklearn.preprocessing import PolynomialFeatures

quadratico = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_treino_quad = quadratico.fit_transform(
    X_treino.reshape(-1, 1)
)

print(X_treino_quad[:5])
```
