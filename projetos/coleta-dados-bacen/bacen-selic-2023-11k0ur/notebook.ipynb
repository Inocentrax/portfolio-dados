## Desenvolvimento


```python
import json
import requests
```

```python
url = 'https://api.bcb.gov.br/dados/serie/bcdata.sgs.11/dados?formato=json&dataInicial=01/01/2023&dataFinal=31/01/2023'
```

```python
resposta = requests.get(url)
```

```python
if resposta.status_code == 200:
  dados_json = resposta.json()
  print(json.dumps(dados_json, indent=4, ensure_ascii=False))
```
