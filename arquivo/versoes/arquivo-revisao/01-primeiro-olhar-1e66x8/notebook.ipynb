## 4. A pergunta do módulo

A pergunta que vamos perseguir por 40 horas é:

> **A sala vai estar ocupada daqui a meia hora?**

Antes de qualquer modelo, vale saber com que frequência ela está ocupada.


```python
taxa = df.ocupada.mean()
print(f'A sala está ocupada em {taxa:.1%} das leituras.')
print(f'Ou seja: está vazia em {1 - taxa:.1%} das leituras.')
```

## 5. Por que `movimento` não resolve

A tentação é óbvia: já existe um sensor de movimento. Por que não usar ele como resposta
e pronto?

A célula abaixo compara as duas colunas.


```python
cruz = pd.crosstab(df.movimento, df.ocupada,
                   rownames=['movimento detectado'],
                   colnames=['sala ocupada'])
print(cruz)
print()

# das vezes em que a sala ESTAVA ocupada, quantas o sensor detectou?
pegou = df.loc[df.ocupada == 1, 'movimento'].mean()
print(f'Com a sala ocupada, o sensor detectou movimento em {pegou:.1%} das leituras.')
print(f'Ou seja: deixou passar {1 - pegou:.1%}.')
print()

# e quando a sala estava vazia?
falso = df.loc[df.ocupada == 0, 'movimento'].mean()
print(f'Com a sala vazia, o sensor acusou movimento em {falso:.1%} das leituras.')
```
