## Desenvolvimento


Imagine que um sistema de segurança registrou sete tentativas de acesso.

```python
tentativas = [1, 0, 1, 1, 0, 0, 1]
```

Precisamos criar um programa que percorra essa lista usando while e conte os acessos autorizados e negados.

```python
i = 0
autorizados = 0
nao_autorizados = 0
while i < len(tentativas):
  if tentativas[i]==1:
    autorizados+=1
  if tentativas[i]==0:
    nao_autorizados+=1
  i+=1

print(f'Tentativas autorizadas:{autorizados}')
print(f'Tentativas não autorizadas: {nao_autorizados}')
```

```python
def analisar_acessos(tentativas):
  i = 0
  autorizados = 0
  nao_autorizados = 0
  invalido = 0
  while i < len(tentativas):
    if tentativas[i]==1:
      autorizados+=1
    if tentativas[i]==0:
      nao_autorizados+=1
    if tentativas[i]!=1 and tentativas[i]!=0:
      invalido+=1
    i+=1
  return autorizados, nao_autorizados, invalido

total_autorizados, total_negados, invalido = analisar_acessos(tentativas)#Desacoplamento de tuplas
print(f'Total autorizados:{total_autorizados}, Total Negados: {total_negados}, Total : {invalido}')#primeiro valor da tupla, na primeira variavel, segundo valor na segunda
```

o return devolve o resultado para a parte que chamou a função

Em Python, assert é uma instrução utilizada para verificar se determinada condição é verdadeira.

```python
try:
    assert 2+3 == 5
except AssertionError as erro:
    print("Exemplo didático de assert não atendido:", erro)
```

```python
try:
    tentativas_teste = [1, 0, 2, 1, 3, 0, 5]
    assert analisar_acessos(tentativas_teste) == (2,2,3)
except AssertionError as erro:
    print("Exemplo didático de assert não atendido:", erro)
```

# Dia 2

```python
tentativas = [1, 0, 1, 1, 0, 0, 1]
tentativas2 = [1, 0, 2, 1, 3, 0, 5]
```

```python
def validar_tentativa(valor):
  if valor == 0:
    return 'negado'
  elif valor==1:
    return 'autorizado'
  else:
    return 'invalido'
```

```python
def analisar_acessos(tentativas):
  autorizados = 0
  nao_autorizados = 0
  invalidos = 0
  i = 0

  while i < len(tentativas):
    classificacao = validar_tentativa(tentativas[i])# Puxou o resultado de uma função dentro do enquanto

    if classificacao=='autorizado': #comparou resultados da outra função como validação
      autorizados+=1#acionando o contador da variavel
    elif classificacao=='negado':
      nao_autorizados+=1
    else:
      invalidos+=1

    i+=1

  return autorizados, nao_autorizados, invalidos # a função cria tupla com 3 valores
```

* fizemos na ultima aula mas não dentro da função(chamada de função)

```python
total_autorizados, total_negados, total_invalidos = analisar_acessos(tentativas)# usa os três valores da tupla
print("Autorizados:", total_autorizados)
print("Negados:", total_negados)
print("Inválidos:", total_invalidos)
```

```python
try:
    assert analisar_acessos(tentativas) == (4,3,0)
except AssertionError as erro:
    print("Exemplo didático de assert não atendido:", erro)
```

```python
total_autorizados, total_negados, total_invalidos = analisar_acessos(tentativas2)
print("Autorizados:", total_autorizados)
print("Negados:", total_negados)
print("Inválidos:", total_invalidos)
```

```python
try:
    assert analisar_acessos(tentativas2) == (2,2,3)
except AssertionError as erro:
    print("Exemplo didático de assert não atendido:", erro)
```
